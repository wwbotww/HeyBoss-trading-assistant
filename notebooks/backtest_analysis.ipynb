# M2 双动量回测结果分析

本 notebook 只读取 `scripts/run_backtest.py` 已生成的报告，不参与生产策略或订单执行。

```python
import json
from pathlib import Path

import pandas as pd

REPORT_DIRECTORY = Path("../reports/backtests/请替换为运行编号")
summary = json.loads((REPORT_DIRECTORY / "summary.json").read_text(encoding="utf-8"))
fills = pd.read_csv(REPORT_DIRECTORY / "fills.csv", parse_dates=["timestamp_utc"])
returns = pd.read_csv(REPORT_DIRECTORY / "returns.csv", parse_dates=["timestamp_utc"])
summary
```

## 成交与费用检查

```python
fills.groupby(["instrument_id", "direction"]).agg(
    fills=("trade_id", "count"),
    quantity=("quantity", "sum"),
    commission=("commission", "sum"),
).sort_index()
```

## 权益与回撤检查

```python
returns["drawdown"] = returns["equity"] / returns["equity"].cummax() - 1
returns[["timestamp_utc", "equity", "cash", "daily_return", "drawdown"]].tail(20)
```
